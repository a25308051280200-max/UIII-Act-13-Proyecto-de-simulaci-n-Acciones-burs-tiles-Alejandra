###  Acciones bursátiles

```python
# ==========================================
# PROGRAMACIÓN ORIENTADA A OBJETOS
# TEMA: ACCIONES BURSÁTILES INTERACTIVO
# ==========================================

class Acciones_Bursatiles:

    # CONSTRUCTOR
    def __init__(self, nombre_empresa, precio_accion,
                 cantidad_acciones, sector_empresa,
                 ganancia_total):

        self.nombre_empresa = nombre_empresa
        self.precio_accion = precio_accion
        self.cantidad_acciones = cantidad_acciones
        self.sector_empresa = sector_empresa
        self.ganancia_total = ganancia_total

    # ======================================
    # FUNCIÓN COMPRAR ACCIONES
    # ======================================
    def comprar_acciones(self):

        cantidad = int(input("¿Cuántas acciones deseas comprar?: "))

        self.cantidad_acciones += cantidad

        print("✅ Compra realizada correctamente")
        print("Acciones actuales:",
              self.cantidad_acciones)

    # ======================================
    # FUNCIÓN VENDER ACCIONES
    # ======================================
    def vender_acciones(self):

        cantidad = int(input("¿Cuántas acciones deseas vender?: "))

        if cantidad <= self.cantidad_acciones:

            self.cantidad_acciones -= cantidad

            print("✅ Venta realizada correctamente")
            print("Acciones restantes:",
                  self.cantidad_acciones)

        else:
            print("❌ No tienes suficientes acciones")

    # ======================================
    # FUNCIÓN MOSTRAR INFORMACIÓN
    # ======================================
    def mostrar_informacion(self):

        print("\n===== INFORMACIÓN BURSÁTIL =====")
        print("Empresa:", self.nombre_empresa)
        print("Precio por acción:", self.precio_accion)
        print("Cantidad de acciones:",
              self.cantidad_acciones)
        print("Sector:", self.sector_empresa)
        print("Ganancia total:",
              self.ganancia_total)

    # ======================================
    # FUNCIÓN CALCULAR INVERSIÓN
    # ======================================
    def calcular_inversion(self):

        inversion = (self.precio_accion *
                     self.cantidad_acciones)

        print("💰 Inversión total:", inversion)

    # ======================================
    # FUNCIÓN ACTUALIZAR PRECIO
    # ======================================
    def actualizar_precio(self):

        nuevo_precio = float(
            input("Ingresa el nuevo precio de la acción: ")
        )

        self.precio_accion = nuevo_precio

        print("📈 Precio actualizado correctamente")
        print("Nuevo precio:",
              self.precio_accion)


# ==========================================
# DATOS INGRESADOS POR EL USUARIO
# ==========================================

print("📊 SISTEMA DE ACCIONES BURSÁTILES 📊")

nombre_empresa = input("Nombre de la empresa: ")
precio_accion = float(input("Precio de la acción: "))
cantidad_acciones = int(input("Cantidad de acciones: "))
sector_empresa = input("Sector de la empresa: ")
ganancia_total = float(input("Ganancia total: "))

# CREAR OBJETO
empresa1 = Acciones_Bursatiles(
    nombre_empresa,
    precio_accion,
    cantidad_acciones,
    sector_empresa,
    ganancia_total
)

# ==========================================
# MENÚ INTERACTIVO
# ==========================================

while True:

    print("\n===== MENÚ =====")
    print("1. Comprar acciones")
    print("2. Vender acciones")
    print("3. Mostrar información")
    print("4. Calcular inversión")
    print("5. Actualizar precio")
    print("6. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        empresa1.comprar_acciones()

    elif opcion == "2":
        empresa1.vender_acciones()

    elif opcion == "3":
        empresa1.mostrar_informacion()

    elif opcion == "4":
        empresa1.calcular_inversion()

    elif opcion == "5":
        empresa1.actualizar_precio()

    elif opcion == "6":
        print("👋 Saliendo del sistema...")
        break

    else:
        print("❌ Opción inválida")
```
